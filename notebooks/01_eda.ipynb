# Dataset Observations

- Dataset contains 284,807 transactions
- Total features: 31
- No missing values present
- Most features are numerical
- Dataset is clean and ready for analysis
- Features V1–V28 are PCA-transformed for privacy
- 'Class' is the target variable

```python
df['Class'].value_counts()
```

```python
#Visualize Fraud Distribution
sns.countplot(x='Class', data=df)

plt.title("Fraud vs Non-Fraud Transactions")

plt.show()
```

```python
#Fraud Percentage
fraud_percentage = (df['Class'].value_counts()[1] / len(df)) * 100

print(f"Fraud Percentage: {fraud_percentage:.4f}%")
```

```python
#Transaction Amount Analysis
df['Amount'].describe()
```

```python
#Amount Distribution Histogram
plt.figure(figsize=(10,5))

sns.histplot(df['Amount'], bins=50)

plt.title("Transaction Amount Distribution")

plt.show()
```

```python
#Compare Fraud vs Normal Amounts
fraud = df[df['Class'] == 1]
normal = df[df['Class'] == 0]

print("Fraud Mean Amount:", fraud['Amount'].mean())

print("Normal Mean Amount:", normal['Amount'].mean())
```

```python
#Boxplot Comparison
sns.boxplot(x='Class', y='Amount', data=df)

plt.title("Amount Distribution by Transaction Type")

plt.show()
```

```python
#Correlation Heatmap
plt.figure(figsize=(20,15))

sns.heatmap(df.corr(), cmap='coolwarm')

plt.title("Correlation Heatmap")

plt.show()
```
